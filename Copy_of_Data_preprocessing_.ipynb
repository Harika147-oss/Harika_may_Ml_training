```python
# Data Preprocessing using NumPy & Pandas
import numpy as np
import pandas as pd
```

```python
# Load Dataset
df = pd.read_csv("student_scores.csv")
print("Dataset Loaded Successfully\n")
```

```python
# Part 1 — NumPy Operations
# Convert math_score column to numeric
df['math_score'] = pd.to_numeric(df['math_score'], errors='coerce')
```

```python
# Create NumPy array
math_scores = df['math_score'].dropna().to_numpy()
print("Math Score Array:")
print(math_scores)
```

```python
# Mean, Median, Max, Min
print("\nStatistics:")
print("Mean:", np.mean(math_scores))
print("Median:", np.median(math_scores))
print("Maximum:", np.max(math_scores))
print("Minimum:", np.min(math_scores))
```

```python
# Normalize scores
normalized_scores = (
    (math_scores - np.min(math_scores)) /
    (np.max(math_scores) - np.min(math_scores))
)
print("\nNormalized Scores:")
print(normalized_scores)
```

```python
#Part 2 — Pandas Exploration
# First 5 rows
print("\nFirst 5 Rows:")
print(df.head())
```

```python
# Data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Students with attendance below 70%
df['attendance'] = pd.to_numeric(df['attendance'], errors='coerce')
low_attendance = df[df['attendance'] < 70]
print("\nStudents with Attendance Below 70%:")
print(low_attendance)
```

```python
# Part 3 — Data Preprocessing
# Identify missing values
print(df.isnull().sum())
# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
# Fill numeric columns with mean
numeric_cols = df.select_dtypes(include=np.number).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())
# Fill categorical columns with mode
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# Convert exam_date to datetime format
df['exam_date'] = pd.to_datetime(df['exam_date'], errors='coerce')
print("\nExam Date Converted:")
print(df['exam_date'])
```

```python
# Detect and remove outliers using IQR method
def remove_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    return data[
        (data[column] >= lower_limit) &
        (data[column] <= upper_limit)
    ]
```

```python
# Remove outliers
df = remove_outliers(df, 'math_score')
df = remove_outliers(df, 'science_score')
print("\nDataset After Removing Outliers:")
print(df)
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
print("\nDataset After Removing Duplicates:")
print(df)
```

```python
# Part 4 — Data Analysis
# Create average_score column
df['average_score'] = (
    df['math_score'] + df['science_score']
) / 2
print("\nAverage Score:")
print(df[['math_score', 'science_score', 'average_score']])
```

```python
# Top 5 students
top_students = df.sort_values(
    by='average_score',
    ascending=False
).head(5)
print("\nTop 5 Students:")
print(top_students)
```

```python
# Correlation between attendance and average score
correlation = df['attendance'].corr(df['average_score'])
print("\nCorrelation Between Attendance and Average Score:")
print(correlation)
```

```python
# Group by gender
gender_average = df.groupby('gender')[
    ['math_score', 'science_score', 'average_score']
].mean()
print("\nAverage Marks by Gender:")
print(gender_average)
```
